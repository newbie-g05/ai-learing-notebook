```python
#类的定义 ----> 命名:每个单词的首字母大写,单词之间没有分隔符
# 定义类
class Person:
    # 构造方法：创建对象的时候自动执行，初始化对象属性
    def __init__(self, name, age):
        # self 代表当前这个对象本身，必须写在第一个参数
        self.name = name   # 对象属性：把传入的name存到对象身上
        self.age = age

    # 对象方法，第一个参数必须是self
    def say_hello(self):
        print(f"你好，我叫{self.name}，今年{self.age}岁")

# 创建对象（实例化），会自动调用 __init__
p1 = Person("张三", 20)
p2 = Person("李四", 22)

# 访问对象属性
print(p1.name)
print(p1.age)

# 调用对象方法
p1.say_hello()
p2.say_hello()
print(p1.__dict__) #-----> __dict__:是自定义类实例的一个特殊属性，用于以字典形式存储对象的属性
```

```python
class Student:
    def __init__(self,name,age,major):
        self.name = name
        self.age = age
        self.major = major

    def introduction(self):
        print(f"我的名字是{self.name},年龄{self.age},专业是{self.major}")
me = Student("GYF",21,"信息与计算科学")
me.introduction()
```

```python
#两数之和
class Solution:
    def twoSum(self, nums: list[int], target: int) -> list[int]:
        di = {}
        for idex,num in enumerate(nums):
            need = target - num
            if need in di:
                return [di[need],idex]
            di[num] = idex
        return []
s = Solution()
res = s.twoSum([1,1,2,3,5,8],13)
print(res)
```
