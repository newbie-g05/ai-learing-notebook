```python
#函数
def circle_area_len(r):#如果返回值有多个，之间用逗号隔开------>多个返回值会封装到元组之中
    #函数说明文档
    """
    根据圆的半径计算面积和周长
    ：param r: 半径
    ：return:圆的面积,圆的周长
    """
    return 3.14 * r * r,round(2 * 3.14 * r, 1)   #round(a,b)----> a保留b位小数
al = circle_area_len(10)
print(al,type(al))
area,len0 = circle_area_len(10)  #元组解包
print(area,len0)
```

```python
def add_item():
    name = input("请输入商品名称:")
    price1 = float(input("请输入商品价格:"))
    num1 = int(input("请输入商品数量:"))
    if name in shopping_cart:
        print("商品已存在,请重新选择")
    else:
        shopping_cart[name] = {"price":price1,"num":num1}
        print("商品已添加")
def query_items():
    if len(shopping_cart) == 0:
        print("购物车暂无商品")
    else:
        for k,v in shopping_cart.items():
            print(f"商品名称:{k} 价格:{v['price']} 数量:{v['num']}")
def remove_item():
    name = input("请输入要删除的商品名称:")
    if name not in shopping_cart:
        print("商品不存在,请重新选择")
    else:
        del shopping_cart[name]
        print("商品已删除")      
def revise_item():
    name = input("请输入要修改的商品名称:")
    if name not in shopping_cart:
        print("商品不存在,请重新选择")
        return
    price1 = float(input("请输入要修改的商品价格:"))
    num1 = int(input("请输入要修改的商品数量:"))
    shopping_cart[name] = {"price":price1,"num":num1}
    print("商品已修改")
shopping_cart = {}
while True:
    print("""
    ======购物车管理系统======
    *      1.添加购物车      *
    *      2.修改购物车      *
    *      3.删除购物车      *
    *      4.查询购物车      *
    *      5.退出购物车      *
    """)
    ch = int(input("请输入选项(1-5):"))
    match ch:
        case 1:
            add_item()
        case 2:
            revise_item()
        case 3:
            remove_item()
        case 4:
            query_items()
        case 5:
            break
        case _:
            print("输入错误,请重新输入:")
```
