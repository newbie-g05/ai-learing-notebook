```python
#函数
def circle_area_len(r):#如果返回值有多个，之间用逗号隔开------>多个返回值会封装到元组之中
    #函数说明文档
    """
    根据圆的半径计算面积和周长
    ：param r: 半径
    ：return:圆的面积,圆的周长
    """
    return 3.14 * r * r,round(2 * 3.14 * r, 1)   #round(a,b)----> a保留b位小数
al = circle_area_len(10)
print(al,type(al))
area,len0 = circle_area_len(10)  #元组解包
print(area,len0)
```

```python
#函数（不定长参数）----> (位置参数*args --> 元组)
def calc_data(*args):
    min_data = min(args)
    max_data = max(args)
    return min_data,max_data
data = (25,36,12,27)
print(calc_data(27,35,12,9))
print(calc_data(*data))
```

```python
#函数（不定长参数）----> (关键字参数**kwargs --> 字典)
def calc_data(*args,**kwargs):
    min_data = min(args)
    max_data = max(args)
    avg_data = sum(args)/len(args)
    if kwargs.get("round") is not None:
        avg_data = round(avg_data,kwargs.get("round"))
    if kwargs.get("print"):
        print(f"最小值:{min_data} 最大值:{max_data} 平均值:{avg_data}")
    return min_data,max_data
data = (25,36,12,27)
print(calc_data(27,35,12,9,round = 2,print = True))
print(calc_data(*data))
```

```python
def factorial(num):#递归调用一定得有终结点
    if num == 1:
        return 1
    else:    
        return num * factorial(num-1)
print(factorial(3))
```

```python
def add_item():
    name = input("请输入商品名称:")
    price1 = float(input("请输入商品价格:"))
    num1 = int(input("请输入商品数量:"))
    if name in shopping_cart:
        print("商品已存在,请重新选择")
    else:
        shopping_cart[name] = {"price":price1,"num":num1}
        print("商品已添加")
def query_items():
    if len(shopping_cart) == 0:
        print("购物车暂无商品")
    else:
        for k,v in shopping_cart.items():
            print(f"商品名称:{k} 价格:{v['price']} 数量:{v['num']}")
def remove_item():
    name = input("请输入要删除的商品名称:")
    if name not in shopping_cart:
        print("商品不存在,请重新选择")
    else:
        del shopping_cart[name]
        print("商品已删除")      
def revise_item():
    name = input("请输入要修改的商品名称:")
    if name not in shopping_cart:
        print("商品不存在,请重新选择")
        return
    price1 = float(input("请输入要修改的商品价格:"))
    num1 = int(input("请输入要修改的商品数量:"))
    shopping_cart[name] = {"price":price1,"num":num1}
    print("商品已修改")
shopping_cart = {}
while True:
    print("""
    ======购物车管理系统======
    *      1.添加购物车      *
    *      2.修改购物车      *
    *      3.删除购物车      *
    *      4.查询购物车      *
    *      5.退出购物车      *
    """)
    ch = int(input("请输入选项(1-5):"))
    match ch:
        case 1:
            add_item()
        case 2:
            revise_item()
        case 3:
            remove_item()
        case 4:
            query_items()
        case 5:
            break
        case _:
            print("输入错误,请重新输入:")
```

```python
import datetime

def write_note():
    # 获取用户输入笔记
    text = input("请输入笔记内容：")
    # 获取当前时间戳
    now = datetime.datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    content = f"[{now}] {text}\n"
    # 追加写入文件
    with open("notes.txt", "a", encoding="utf-8") as f:
        f.write(content)
    print("笔记保存成功！")

def read_notes():
    try:
        with open("notes.txt", "r", encoding="utf-8") as f:
            data = f.read()
        print("===== 所有笔记 =====")
        print(data)
    except FileNotFoundError:
        print("错误：notes.txt 文件不存在，先新增笔记再读取！")

# 程序交互入口
if __name__ == "__main__":
    while True:
        print("\n1. 新增笔记")
        print("2. 查看全部笔记")
        print("3. 退出")
        op = input("请选择功能序号：")
        if op == "1":
            write_note()
        elif op == "2":
            read_notes()
        elif op == "3":
            print("程序结束")
            break
        else:
            print("输入无效，请输入1/2/3")
```
