```python
shopping_cart = {}
while True:
    print("""
    ======购物车管理系统======
    *      1.添加购物车      *
    *      2.修改购物车      *
    *      3.删除购物车      *
    *      4.查询购物车      *
    *      5.退出购物车      *
    """)
    ch = input("请输入选项(1-5):")
    match ch:
        case "1":
            name = input("请输入商品名称:")
            price1 = float(input("请输入商品价格:"))
            num1 = int(input("请输入商品数量:"))
            if name in shopping_cart:
                print("商品已存在,请重新选择")
            else:
                shopping_cart[name] = {"price":price1,"num":num1}
                print("商品已添加")
        case "2":
            name = input("请输入要修改的商品名称:")
            if name not in shopping_cart:
                print("商品不存在,请重新选择")
                continue
            price1 = float(input("请输入要修改的商品价格:"))
            num1 = int(input("请输入要修改的商品数量:"))
            shopping_cart[name] = {"price":price1,"num":num1}
            print("商品已修改")
        case "3":
            name = input("请输入要删除的商品名称:")
            if name not in shopping_cart:
                print("商品不存在,请重新选择")
            else:
                del shopping_cart[name]
                print("商品已删除")
        case "4":
            if len(shopping_cart) == 0:
                print("购物车暂无商品")
            else:
                for k,v in shopping_cart.items():
                    print(f"商品名称:{k} 价格:{v['price']} 数量:{v['num']}")
        case "5":
            break
        case _:
            print("输入错误,请重新输入:")
```

```python
def add_item():
    name = input("请输入商品名称:")
    price1 = float(input("请输入商品价格:"))
    num1 = int(input("请输入商品数量:"))
    if name in shopping_cart:
        print("商品已存在,请重新选择")
    else:
        shopping_cart[name] = {"price":price1,"num":num1}
        print("商品已添加")
def query_items():
    if len(shopping_cart) == 0:
        print("购物车暂无商品")
    else:
        for k,v in shopping_cart.items():
            print(f"商品名称:{k} 价格:{v['price']} 数量:{v['num']}")
def remove_item():
    name = input("请输入要删除的商品名称:")
    if name not in shopping_cart:
        print("商品不存在,请重新选择")
    else:
        del shopping_cart[name]
        print("商品已删除")      
def revise_item():
    name = input("请输入要修改的商品名称:")
    if name not in shopping_cart:
        print("商品不存在,请重新选择")
        return
    price1 = float(input("请输入要修改的商品价格:"))
    num1 = int(input("请输入要修改的商品数量:"))
    shopping_cart[name] = {"price":price1,"num":num1}
    print("商品已修改")
shopping_cart = {}
while True:
    print("""
    ======购物车管理系统======
    *      1.添加购物车      *
    *      2.修改购物车      *
    *      3.删除购物车      *
    *      4.查询购物车      *
    *      5.退出购物车      *
    """)
    ch = int(input("请输入选项(1-5):"))
    match ch:
        case 1:
            add_item()
        case 2:
            revise_item()
        case 3:
            remove_item()
        case 4:
            query_items()
        case 5:
            break
        case _:
            print("输入错误,请重新输入:")
```
