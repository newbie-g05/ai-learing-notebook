```python
#字符串常用方法str
s = " Hello Python "

s.strip()       # 去掉首尾空格
s.lower()       # 全部小写
s.upper()       # 全部大写
s.split(" ")    # 按空格切割，返回列表
s.replace("H","A") # 替换
s.find("Py")    # 找下标，找不到返回-1
"Py" in s       # 判断是否包含，返回True/False

len(s)          # 获取字符串长度
```

```python
t = (2,5,2,8)

len(t)          # 长度
t.count(2)      # 统计元素出现次数
t.index(5)      # 返回元素下标，找不到报错

# 拼接、复制，生成新元组
t1 = (1,2) + (3,4)
t2 = (10,) * 3

# in 判断是否存在
3 in t1
```

```python
s = {10,20,30}
m = {30,40,50}
s.add(40)        # 添加单个元素
s.update([50,60])# 批量添加

s.remove(20)     # 删除元素，不存在直接报错
s.discard(99)    # 删除，不存在不报错
s.pop()          # 随机删掉一个元素并返回所删除的元素
s.clear()        # 清空集合
s.difference(m)  #差集difference()
s.union(m)       #并集union()
s.intersection(m)#交集intersection()
#集合推导式------> {要添加的数据 for i in 集合 if 条件}
```

```python
# 字典：key:value 键值对，key 必须不可变（字符串、数字、元组），不能用列表做 key，value 可以任意类型。
# 定义
d = {"name":"张三", "age":18, "hobby":["跑步","看书"]}

# 查询
print(d["name"])         # 通过key取值，key不存在直接报错
print(d.get("age"))      # get取值，key不存在返回None，不会报错
print(d.get("score",0))  # 找不到给默认值0

# 新增 / 修改：key存在就修改，不存在就新增
d["gender"] = "男"
d["age"] = 19

# 删除
del d["gender"]               # 删除指定键值对
val = d.pop("age")            # 删除并拿到被删除的值
d.clear()                     # 清空整个字典

# 遍历key和value（最常用）
for k,v in d.items():
    print(k, v)
```

```python
shopping_cart = {}
while True:
    print("""
    ======购物车管理系统======
    *      1.添加购物车      *
    *      2.修改购物车      *
    *      3.删除购物车      *
    *      4.查询购物车      *
    *      5.退出购物车      *
    """)
    ch = input("请输入选项(1-5):",)
    match ch:
        case "1":
            name = input("请输入商品名称:",)
            price1 = float(input("请输入商品价格:",))
            num1 = int(input("请输入商品数量:",))
            if name in shopping_cart:
                print("商品已存在,请重新选择")
            else:
                shopping_cart[name] = {"price":price1,"num":num1}
                print("商品已添加")
        case "2":
            name = input("请输入要修改的商品名称:",)
            if name not in shopping_cart:
                print("商品不存在,请重新选择")
                continue
            price1 = float(input("请输入要修改的商品价格:",))
            num1 = int(input("请输入要修改的商品数量:",))
            shopping_cart[name] = {"price":price1,"num":num1}
            print("商品已修改")
        case "3":
            name = input("请输入要删除的商品名称:",)
            if name not in shopping_cart:
                print("商品不存在,请重新选择")
            else:
                del shopping_cart[name]
                print("商品已删除")
        case "4":
            if len(shopping_cart) == 0:
                print("购物车暂无商品")
            else:
                for k,v in shopping_cart.items():
                    print(f"商品名称:{k} 价格:{v['price']} 数量:{v['num']}")
        case "5":
            break
        case _:
            print("输入错误,请重新输入:")
```
