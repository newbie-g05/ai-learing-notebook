```python
#print()输出
#input()输入
#// ---> 整除，结果为整数;** ---> 幂指数，a ** b表示a的b次方
#逻辑运算符:and,or,not

a = [11,22,33,44]
print(a[0])      # 取第1个：11
print(a[-1])     # 取最后一个：44
a.append(33)         # 末尾追加：[11,22,33]
a.insert(1,99)       # 指定索引位置插入，索引1插入99 → [11,99,22,33]
print(len(a))    # len() 获取列表长度 4
```
