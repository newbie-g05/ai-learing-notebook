### 列表：增、删、改、查

```python
a = [11,22,33,22]
print(a[-1])     # 取最后一个：44
print(len(a))    # len() 获取列表长度 4
a.remove(22)   # 删除第一个匹配的元素，删掉第一个22
del a[0]       # 根据索引删除
val = a.pop()  # 删除末尾元素，同时返回被删掉的值
# a.clear()      # 清空整个列表
print(a)
```

```python
a = [11,22,33,44]
print(a[0])      # 取第1个：11
print(a[-1])     # 取最后一个：44
a.append(33)         # 末尾追加：[11,22,33]
a.insert(1,99)       # 指定索引位置插入，索引1插入99 → [11,99,22,33]
print(len(a))    # len() 获取列表长度 4
```

## if 判断语句
### 注意 Python 靠缩进（4 空格）代表代码块，条件后面加冒号 :

```python
age = int(input())
if age < 18:
    print("未成年")
elif age == 18:
    print("刚成年")
else:
    print("成年")
```

### 配合 range，遍历数字

```python
# range(5) → 0,1,2,3,4
for i in range(5):
    print(i)

# range(起始,结束,步长)
for i in range(1,10,2):
    print(i) #1 3 5 7 9
```

### 同时拿到索引 + 元素 enumerate

```python
lst = [100,200,300]
for idx,val in enumerate(lst):
    print(idx, val)
```
