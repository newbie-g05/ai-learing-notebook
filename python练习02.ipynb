### 列表：增、删、改、查

```python
a = [11,22,33,22]
print(a[-1])     # 取最后一个：44
print(len(a))    # len() 获取列表长度 4
a.remove(22)   # 删除第一个匹配的元素，删掉第一个22
del a[0]       # 根据索引删除
val = a.pop()  # 删除末尾元素，同时返回被删掉的值
# a.clear()      # 清空整个列表
print(a)
```

```python
a = [11,22,33,44]
#切片a[开始索引:结束索引:步长] --->左闭右开
print(a[0:2:1])
print(a[:3:])#默认开始索引为0，步长为1,且第二个":"可省略,即print(a[:3])
print(a[0])      # 取第1个：11
print(a[-1])     # 取最后一个：44
a.append(33)         # 末尾追加：[11,22,44，33]
a.insert(1,99)       # 指定索引位置插入，索引1插入99 → [11,99,22,44，33]
print(len(a))    # len() 获取列表长度 4
```

```python
#python随机数需先写import random
#随机整数random.randint(a,b)
import random
answer = random.randint(1,100)
print("猜数字(1-100)")
i = 0
while True:
    i += 1
    if i > 7:
        print(f"游戏结束,你输了!\n猜了{i-1}次,正确答案是{answer}")
        break
    a = int(input("请输入数字",))
    if a > answer:
        print("猜大了")
    elif a < answer:
        print("猜小了")
    else:
        print(f"猜对了，猜了{i}次!")
        break
```

```python
list1 = [0,1,2,3,4]
list2 = [5,6,7,8,9]
#解包:将列表这一类容器解开成一个一个独立的元素
#将多个值合并到一个容器
list3 = [*list1,*list2]
print(f"{list3}")
#列表推导式[变量/关系式 for i in 序列/列表 if 条件]
list4 = [i**2 for i in range(1,6)]
print(f"{list4}")
list5 = [i**2 for i in list4 if i**2 % 2 == 0]
print(f"{list5}")
```
